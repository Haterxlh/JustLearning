### 思路
使用优先队列（heapq）模拟 + 哈希表存值

````python
import sys
import heapq
data = sys.stdin.buffer.read().split()
it = iter(data)

K = int(next(it))
N = int(next(it))

cache = []
vals = {}

def pout(s):
    sys.stdout.write(f"{s}\n")

for _ in range(N):
    op = str(next(it).decode("utf-8"))
    if op == "ADD":
        pos = int(next(it))
        score = float(next(it))
        val = [[float(next(it)) for i in range(4)] for j in range(2)]
        heapq.heappush(cache, (score, pos))
        vals[pos] = {
            "score": score,
            "val": val
        }
        if len(cache) > K:
            old_score, old_pos = heapq.heappop(cache)
            pout(f"PRUNED {old_pos}")
            del vals[old_pos]
    else: # QUERY
        pout(f"{len(cache)}")
        poss = list(vals.keys())
        for pos in poss:
            item = vals[pos]
            score = item["score"]
            val = item["val"]
            pout(f"{pos} {score}")
            pout(" ".join(map(str, val[0])))
            pout(" ".join(map(str, val[1])))

'''
提交结果：答案正确 运行时间：1499ms 占用内存：78284KB 使用语言： Python 3 用例通过率：100.00%
'''
````

### 注意事项（踩到的坑）：
1. 看题不仔细，将“先插入再判断是否超容”看成了“先判断是否已满，再插入”
2. 使用 sys.stdin.buffer.read().split() 读取输入时，如果是字符串，还需要使用 decode 将 byte 还原成 utf-8
    ```python
    op = str(next(it))   # op 是 "b'ADD'"
    if op == "ADD":      # False
    ```
3. heapq的操作不是很熟练
    ```python
    item = heapq.heappop(h)
    heapq.heappush(h, item)
    ```
