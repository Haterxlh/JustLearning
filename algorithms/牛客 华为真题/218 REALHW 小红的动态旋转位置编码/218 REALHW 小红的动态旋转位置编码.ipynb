### 218 REALHW 小红的动态旋转位置编码

链接：[218 REALHW 小红的动态旋转位置编码](https://www.nowcoder.com/practice/e33f7542d4004724be19f93b40687266?tpId=390&tqId=11745345&sourceUrl=%2Fexam%2Foj%2Fta%3FtpId%3D37)

```python
import sys
from math import pow, cos, sin

data = sys.stdin.buffer.read().split()
it = iter(data)

T, L, D, b, m = float(next(it)), float(next(it)), int(next(it)), float(next(it)), float(next(it))
s = max(1, T / L)
b_prime = b * pow(s, D / (D - 2))
x = [float(next(it)) for _ in range(D)]
theta = [m * pow(b_prime, - 2 * i / D) for i in range(D // 2)]

minv = 5e-5
x_prime = []
for i in range(D // 2):
    cosv = cos(theta[i])
    sinv = sin(theta[i])
    x_2i = x[2 * i]
    x_2i1 =  x[2 * i + 1]
    a = x_2i * cosv - x_2i1 * sinv
    b = x_2i * sinv + x_2i1 * cosv
    x_prime.append(a)
    x_prime.append(b)


sys.stdout.write(" ".join(map(lambda x: f"{x:.4f}" if abs(x) >= minv else "0.0000", x_prime)))

"""
通过全部用例
运行时间
532ms
占用内存
55276KB
"""
```
