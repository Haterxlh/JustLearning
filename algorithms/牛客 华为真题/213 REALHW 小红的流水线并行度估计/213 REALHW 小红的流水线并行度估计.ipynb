### 213 REALHW 小红的流水线并行度估计

````python
import sys
import math
def solve():
    data = sys.stdin.buffer.read().split()
    it = iter(data)
    if len(data) != 4:
        print("-1 -1")
        return
    
    eps = 1e-9

    try:
        L = float(next(it))
        C = float(next(it))
        m = float(next(it))
        H = float(next(it))
    except Exception as e:
        print("-1 -1")
        return

    if L <= 0 or C <= 0 or m <= 0 or H <= 0:
        print("-1 -1")
        return

    if abs(L - int(L)) > eps or abs(C - int(C)) > eps or abs(m - int(m)) > eps or abs(H - int(H)) > eps:
        print("-1 -1")
        return
    
    L = int(L)
    C = int(C)
    m = int(m)
    H = int(H)

    ps = []

    # 找 L 的公约数
    divisors = []
    N = math.ceil(math.sqrt(L) + 1)
    p = 1
    while p <= N:
        if (L % p) == 0:
            divisors.append(p)
            divisors.append(L // p)
        p += 1

    divisors.sort()
    # 检验
    for p in divisors:
        if 1 <= p and p <= H and p >= math.ceil(L / C):
            ps.append((p, 1 - (m / (m + p - 1))))
            break

    if not ps:
        print("-1 -1")
        return
    
    print(f"{ps[0][0]} {ps[0][1]:.4f}")

solve()

"""
通过全部用例
运行时间
40ms
占用内存
4740KB
"""
````

踩坑：

1. 找公约数：
    ```python
    if (L % p) == 0:
        divisors.append(p)
    ```
    没有把 (L // p) 也加进去


优化：

1. math.ceil(L / C) 可以用整数运算 (L + C - 1) // C 避免浮点误差。
2. 输入合法性检查不严谨
    
    ```python
    # 可以直接用 str.isdigit()判断输入的合法性
    # 注意 isdigit() 对 bytes 也适用，但需要先解码或直接用 bytes.isdigit()。
    curin = next(it)
    if not curin.isdigit():
        print("-1 -1")
        return
    ```
