### REALHW223 小红的特征加权推荐

链接: [REALHW223 小红的特征加权推荐](https://www.nowcoder.com/practice/4b6427c717e6483a8ab85377351385b7?tpId=390&tqId=11745384&sourceUrl=%2Fexam%2Foj%2Fta%3FtpId%3D37)

![题目描述](./desc.png)


```python
import sys

data = sys.stdin.buffer.read().split()
it = iter(data)

N, D, K = int(next(it)), int(next(it)), int(next(it))
W = [int(next(it)) for _ in range(D)]
target = [int(next(it)) for _ in range(D)]
features = [[0 for _ in range(D)] for i in range(N)]
idxs = []
for i in range(N):
    idxs.append(int(next(it)))
    for j in range(D):
        features[i][j] = int(next(it))

scores = []

def get_score(a, b, w):
    score = 0
    for j in range(D):
        score += w[j] * (b[j] - a[j]) * (b[j] - a[j])
    return score


for i in range(N):
    b = features[i]
    score = get_score(target, b, W)
    scores.append((score, i))

scores.sort(key=lambda x: (x[0], x[1]))
# sys.stdout.write(" ".join(map(lambda x: idxs[x], scores[:min(K, N)][1])))
res = [idxs[scores[i][1]] for i in range(min(K, N))]
print(*res)

"""
通过全部用例
运行时间
525ms
占用内存
52128KB
"""
```

1. 对于二维的list，采用 scores[:K][1] 的形式等同于 先切片再切片
