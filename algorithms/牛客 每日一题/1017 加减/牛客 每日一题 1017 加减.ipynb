```shell
示例1

输入：

5 3

6 3 20 8 1

输出：

2

说明：
共 3 次操作如下：

第一个数加一。
第二个数加一。
第四个数减一。

数组变成了 7 4 20 7 1 ，共有 2 个相同的数： 7 。

可以证明， 2 为最优解。另外，此上操作并不是唯一的操作。 
```


````python
import sys

data = sys.stdin.buffer.read().split()
it = iter(data)
n, k = int(next(it)), int(next(it))
a = [int(next(it)) for _ in range(n)]
a.append(0)
a.sort()

s = [0 for _ in range(n + 1)]
for i in range(1, n + 1):
    s[i] = s[i - 1] + a[i]

def check(m):
    global a, s, n, k
    # m 表示这个最多元素出现的次数
    # 那么这m个数在a数组中肯定是连续的
    # 枚举这个区间的左端点下标
    for l in range(1, n + 1):
        # 右端点下标
        r = l + m - 1
        if r <= n:
            mid = (l + r) // 2
            # 看看把这个区间内的 m 个数全部变成下标 mid 对应的数 所需要的操作数量是否会超过 k
            # 变 [l, mid] 所需的操作数
            l_op_nums = (mid - l + 1) * a[mid] - (s[mid] - s[l - 1])
            # 变 (mid, r] 所需的操作数
            r_op_nums = (s[r] - s[mid]) - (r - mid) * a[mid]
            if (l_op_nums + r_op_nums) <= k:
                return True
    
    return False

# i, j 表示搜索 m 的左右区间
i = 1
j = n
while i < j:
    m = (i + j + 1) // 2
    if check(m):
        i = m
    else:
        j = m - 1

print(i)

"""
通过全部用例
运行时间
1717ms
占用内存
19788KB
"""
````

踩坑：

1. 二分边界问题，取最大的边界时，不满足要求时右边界变成 mid - 1, mid 的更新时应该是 上取整
2. 前缀和快速计算区间内的操作数
