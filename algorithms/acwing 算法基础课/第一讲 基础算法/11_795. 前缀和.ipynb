```python
import sys
data = sys.stdin.buffer.read().split()
it = iter(data)

n, k = int(next(it)), int(next(it))
a = [0] + [int(next(it)) for _ in range(n)]
s = [0] * (n + 1)
for i in range(1, n + 1): s[i] = s[i - 1] + a[i]

while k:
    k -= 1
    l, r = int(next(it)), int(next(it))
    sys.stdout.write(f"{s[r] - s[l - 1]}\n")
```
