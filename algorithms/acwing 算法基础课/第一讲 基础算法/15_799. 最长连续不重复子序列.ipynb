```python
import sys
from collections import defaultdict

data = sys.stdin.buffer.read().split()
it = iter(data)

n = int(next(it))
a = [int(next(it)) for _ in range(n)]

s = defaultdict(int)

j = 0
res = 0
for i in range(n):
    s[a[i]] += 1
    while s[a[i]] > 1:
        s[a[j]] -= 1
        j += 1
    res = max(res, i - j + 1)
sys.stdout.write(f"{res}\n")
```
