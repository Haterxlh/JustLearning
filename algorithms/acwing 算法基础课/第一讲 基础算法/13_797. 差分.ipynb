```python
import sys
data = sys.stdin.buffer.read().split()
it = iter(data)

n, k = int(next(it)), int(next(it))
a = [0] + [int(next(it)) for _ in range(n)]
d = [0] * (n + 2)

for i in range(1, n + 1): d[i] = a[i] - a[i - 1]

while k:
    k -= 1
    l, r, c = int(next(it)), int(next(it)), int(next(it))
    d[l] += c; d[r + 1] -= c
    
s = [0] * (n + 1)
for i in range(1, n + 1): s[i] = s[i - 1] + d[i]

sys.stdout.write(" ".join(map(str, s[1:])))
```
