```python
import sys
data = sys.stdin.buffer.read().split()
it = iter(data)

str_a = next(it).decode()
b = int(next(it))

a = []
for i in range(len(str_a)):
    a.append(ord(str_a[i]) - ord('0'))

def big_div(a:list[int], b:int):
    c = []
    r = 0
    for i in range(len(a)):
        r *= 10
        r += a[i]
        c.append(r // b)
        r = r % b
    return c, r
    
res, r = big_div(a, b)
while len(res) > 1 and not res[0]:
    res.pop(0)
sys.stdout.write("".join(map(str, res)) + f"\n{r}")
```
