```python
import sys
data = sys.stdin.buffer.read().split()
it = iter(data)

n, m, x = int(next(it)), int(next(it)), int(next(it))
a = [int(next(it)) for _ in range(n)]
b = [int(next(it)) for _ in range(m)]

i = 0
j = m - 1
while a[i] + b[j] != x:
    if a[i] + b[j] > x: j -= 1
    else: i += 1

sys.stdout.write(f"{i} {j}")
```
