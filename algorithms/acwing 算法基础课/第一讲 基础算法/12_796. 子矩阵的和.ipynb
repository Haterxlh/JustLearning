```python
import sys
data = sys.stdin.buffer.read().split()
it = iter(data)

n, m, k = int(next(it)), int(next(it)), int(next(it))
a = [[0] * (m + 1)]  + [[0] + [int(next(it)) for _ in range(m)] for _ in range(n)]
s = [[0] * (m + 1) for _ in range(n + 1)]

for i in range(1, n + 1):
    for j in range(1, m + 1):
        s[i][j] = s[i][j - 1] + s[i - 1][j] - s[i - 1][j - 1] + a[i][j]

while k:
    k -= 1
    x1, y1, x2, y2 = int(next(it)), int(next(it)), int(next(it)), int(next(it))
    sys.stdout.write(f"{s[x2][y2] - s[x1 - 1][y2] - s[x2][y1 - 1] + s[x1 - 1][y1 - 1]}\n")
```
