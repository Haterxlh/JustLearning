````python
import sys
data = sys.stdin.buffer.read().split()
it = iter(data)

str_a = next(it).decode()
str_b = next(it).decode()

a = []
b = []
for i in range(len(str_a) - 1, -1, -1):
    a.append(ord(str_a[i]) - ord('0'))

for i in range(len(str_b) - 1, -1, -1):
    b.append(ord(str_b[i]) - ord('0'))
    
def big_add(a: list, b: list) -> list:
    c = []
    t = 0
    n = max(len(a), len(b))
    for i in range(n):
        if i < len(a): t += a[i]
        if i < len(b): t += b[i]
        c.append(t % 10)
        t = t // 10
    while t:
        c.append(t % 10)
        t = t // 10
    return c
    
res = big_add(a, b)[::-1]
sys.stdout.write("".join(map(str, res)))
````
