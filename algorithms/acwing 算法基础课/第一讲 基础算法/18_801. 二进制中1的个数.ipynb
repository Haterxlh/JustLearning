### v1

```python
import sys
data = sys.stdin.buffer.read().split()
it = iter(data)
n = int(next(it))

ans = []
while n:
    n -= 1
    x = int(next(it))
    res = 0
    while x:
        res += x % 2
        x >>= 1
    ans.append(res)
sys.stdout.write(" ".join(map(str, ans)))
```

注意：

循环中不要每次都 sys.stdout.write(f"{res} ")

这样会很慢

### v2

优化每次 更新x的方法

首先需要知道的是 & 运算符

拿 ``a & b`` 举例：

这个操作是将 a 和 b 的二进制表示中的每一位进行 ``与`` 操作

那么具体的 ``x & 1`` 的话，

对于 1 而言，其二进制表示中除了最低位是1，其他位都是0

所以 ``x & 1`` 会将 x 的最低位取出来

也借此，我们可以通过 ``x & x - 1`` 从而从 x 中减去 x 最低位的 1 对应值减去

```python
import sys
data = sys.stdin.buffer.read().split()
it = iter(data)
n = int(next(it))

ans = []
while n:
    n -= 1
    x = int(next(it))
    res = 0
    while x:
        x = x & x - 1
        res += 1
    ans.append(res)
sys.stdout.write(" ".join(map(str, ans)))
```
