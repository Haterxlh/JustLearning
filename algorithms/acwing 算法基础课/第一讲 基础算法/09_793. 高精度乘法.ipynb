```python
import sys
data = sys.stdin.buffer.read().split()
it = iter(data)

str_a = next(it).decode()
b = int(next(it))

a = []
for i in range(len(str_a) - 1, -1, -1):
    a.append(ord(str_a[i]) - ord('0'))
    
def big_mul(a:list[int], b:int):
    c = []
    t = 0
    for i in range(len(a)):
        t += a[i] * b
        c.append(t % 10)
        t = t // 10
    while t:
        c.append(t % 10)
        t = t // 10
    return c
    
res = big_mul(a, b)
res = res[::-1]
# 去除前导零
while len(res) > 1 and not res[0]:
    res.pop(0)
sys.stdout.write("".join(map(str, res)))
```
