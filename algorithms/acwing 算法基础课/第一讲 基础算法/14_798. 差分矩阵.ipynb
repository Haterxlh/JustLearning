```python
import sys
data = sys.stdin.buffer.read().split()
it = iter(data)

n, m, k = int(next(it)), int(next(it)), int(next(it))
a = [[0] * (m + 1)]  + [[0] + [int(next(it)) for _ in range(m)] for _ in range(n)]
d = [[0] * (m + 2) for _ in range(n + 2)]
for i in range(1, n + 1):
    for j in range(1, m + 1):
        d[i][j] = a[i][j] - a[i][j - 1] - a[i - 1][j] + a[i - 1][j - 1]

while k:
    k -= 1
    x1, y1, x2, y2, c = tuple([int(next(it)) for _ in range(5)])
    d[x1][y1] += c
    d[x2 + 1][y1] -= c
    d[x1][y2 + 1] -= c
    d[x2 + 1][y2 + 1] += c
    

s = [[0] * (m + 1) for _ in range(n + 1)]
for i in range(1, n + 1):
    for j in range(1, m + 1):
        s[i][j] = s[i - 1][j] + s[i][j - 1] - s[i - 1][j - 1] + d[i][j]

sys.stdout.write("\n".join([" ".join(map(str, l[1:])) for l in s[1:]]))
```
