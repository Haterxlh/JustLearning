```python
import sys
data = sys.stdin.buffer.read().split()
it = iter(data)

n, k = int(next(it)), int(next(it))
a = [int(next(it)) for _ in range(n)]

def quick_choose(l:int, r:int, k:int):
    global a
    if l == r: return a[l]
    x = a[(l + r) // 2]; i = l - 1; j = r + 1
    while i < j:
        i += 1
        while a[i] < x: i += 1
        j -= 1
        while a[j] > x: j -= 1
        if i < j: a[i], a[j] = a[j], a[i]
    
    cur_len = j - l + 1
    if cur_len >= k:
        return quick_choose(l, j, k)
    return quick_choose(j + 1, r, k - cur_len)

print(quick_choose(0, n - 1, k))
```

思路：利用快排的“标兵”，“标兵”位置确定后，确定第k大的数在左还是右
