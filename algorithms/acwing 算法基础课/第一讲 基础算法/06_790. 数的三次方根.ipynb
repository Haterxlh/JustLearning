```python
import sys
import math

def solve():
    data = sys.stdin.buffer.read().split()
    it = iter(data)
    x = float(next(it))
    
    def cube_root(x:float, l:float, r:float):
        eps = 1e-9
        while abs(l - r) > eps:
            mid = (l + r) / 2
            if math.pow(mid, 3) > x: r = mid
            else: l = mid
        
        return r
    res = cube_root(x, -10000.0, 10000.0)
    sys.stdout.write(f"{res:.6f}\n")
    
solve()
```

这里说了数据范围是 −10000≤x≤10000

虽然直接用 l=-10000.0, r=10000.0 能过
但是不太优雅，感觉可以后面分别处理 -1≤x≤1 的情况
