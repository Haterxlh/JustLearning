```python
import sys
data = sys.stdin.buffer.read().split()
it = iter(data)

n = int(next(it))
a = [int(next(it)) for _ in range(n)]

def quick_sort(l:int, r:int):
    if l >= r: return
    mid = a[(l + r) // 2]; i = l - 1; j = r + 1
    while i < j:
        i += 1
        while a[i] < mid: i += 1
        j -= 1
        while a[j] > mid: j -= 1
        if i < j: a[i], a[j] = a[j], a[i]
    quick_sort(l, j); quick_sort(j + 1, r)

quick_sort(0, n - 1)
sys.stdout.write(" ".join(map(str, a)))
```

问题1：

为什么初始化 i = l - 1; j = r + 1

> 目的：配合循环中的“先移动，再判断”的写法
> 为什么先移动？因为这样能避免死循环（a[i] a[j] 都 == mid）

问题2：

为什么最后使用j作为分界，而不是用i？

> 原因：这种写法决定了，循环结束后 j 一定是“左半边”（都是<=mid）的最后一个元素
> i 有两种可能，
> 1. i == j , 此时如果分界 [l, i - 1], [i, r], 如果 i == j == l, 就会导致右边的范围无法缩小（[i, r] == [l, r]），而进入死循环
> 2. i == j + 1
