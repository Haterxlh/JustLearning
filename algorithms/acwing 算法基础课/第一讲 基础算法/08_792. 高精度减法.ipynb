```python
import sys
data = sys.stdin.buffer.read().split()
it = iter(data)

str_a = next(it).decode()
str_b = next(it).decode()

def bigger(a:str, b:str):
    if len(a) != len(b): return len(a) > len(b)
    for i in range(len(a)):
        if ord(a[i]) != ord(b[i]):
            return ord(a[i]) > ord(b[i])
    return True 
    
prefix = '' if bigger(str_a, str_b) else '-'

a = []
b = []

for i in range(len(str_a) - 1, -1, -1):
    a.append(ord(str_a[i]) - ord('0'))
for i in range(len(str_b) - 1, -1, -1):
    b.append(ord(str_b[i]) - ord('0'))
    
def big_minus(a:list, b:list):
    """
    默认 a >= b
    """
    c = []
    t = 0
    for i in range(len(a)):
        t += a[i]
        if i < len(b): t -= b[i]
        if t >= 0: c.append(t); t = 0
        else: c.append(10 + t); t = -1
        
    return c
    
res = big_minus(a, b) if bigger(str_a, str_b) else big_minus(b, a)
res = res[::-1]
# 去除前导零
while len(res) > 1 and not res[0]:
    res.pop(0)
sys.stdout.write(f"{prefix}{''.join(map(str, res))}")
```

重点：

1. 字符串数字比较大小
2. 去除前导零
