```python
import sys

def solve():
    data = sys.stdin.buffer.read().split()
    it = iter(data)
    
    n, k = int(next(it)), int(next(it))
    a = [int(next(it)) for _ in range(n)]
    
    def left_search(x:int, l:int, r:int):
        while l < r:
            mid = (l + r) >> 1
            if a[mid] < x: l = mid + 1
            else: r = mid
            
        if a[l] != x: return -1
        return l
    
    def right_search(x:int, l:int, r:int):
        while l < r:
            mid = (l + r + 1) >> 1
            if a[mid] > x: r = mid - 1
            else: l = mid
            
        if a[r] != x: return -1
        return r
        
    while k:
        k -= 1
        x = int(next(it))
        l = left_search(x, 0, n - 1)
        if l == -1:
            sys.stdout.write("-1 -1\n")
        else:
            r = right_search(x, 0, n - 1)
            sys.stdout.write(f"{l} {r}\n")
    
solve()
```

总结：

1. 二分找左边界：mid 下取整，调整 l = mid + 1
2. 二分找右边界：mid 上取整，调整 r = mid - 1
