```python
import sys
data = sys.stdin.buffer.read().split()
it = iter(data)
n, m = int(next(it)), int(next(it))
a = [int(next(it)) for _ in range(n)]
b = [int(next(it)) for _ in range(m)]

i = 0
j = 0

res = "No"
while j < m:
    if b[j] == a[i]: 
        i += 1
        if i == n:
            res = "Yes"
            break
    j += 1

sys.stdout.write(res)
```
