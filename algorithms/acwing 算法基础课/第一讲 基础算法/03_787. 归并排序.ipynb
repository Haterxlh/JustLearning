```python
import sys
data = sys.stdin.buffer.read().split()
it = iter(data)
n = int(next(it))
a = [int(next(it)) for _ in range(n)]

def merge_sort(l:int, r:int):
    if l >= r: return
    mid = (l + r) // 2
    merge_sort(l, mid); merge_sort(mid + 1, r)
    
    tmp = []
    i = l; j = mid + 1
    while i <= mid and j <= r:
        if a[i] <= a[j]: tmp.append(a[i]); i += 1
        else: tmp.append(a[j]); j += 1
        
    while i <= mid: tmp.append(a[i]); i += 1
    while j <= r: tmp.append(a[j]); j += 1
    for idx in range(r - l  + 1): a[l + idx] = tmp[idx]

merge_sort(0, n - 1)
sys.stdout.write(" ".join(map(str, a)))
```
