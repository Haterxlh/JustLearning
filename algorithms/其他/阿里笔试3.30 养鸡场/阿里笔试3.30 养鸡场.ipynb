题目一：
第一题，第一行输入：n,m,k
表示n个养鸡场，m天，每个鸡场增加k只鸡

第二行输入：n个数，分别代表每个鸡场里面鸡的数量。
题意：每天每个鸡场增加k个鸡，每天结束找到最多数量的鸡场，使之减半。

输出：m天后所有鸡场的数量总和

用例输入：3 3 100
100  200  400
输出：925

作者：Bob_linux
链接：https://www.nowcoder.com/discuss/353156241134395392?sourceSSR=search
来源：牛客网

```python
import sys
import heapq

data = sys.stdin.read().split()
it = iter(data)

n, m, k = int(next(it)), int(next(it)), int(next(it))
a = [-int(next(it)) for _ in range(n)]

heapq.heapify(a)

for i in range(m):
    base = (i + 1) * k
    # 每天取出最多的鸡
    maxv = heapq.heappop(a)
    # 计算剩下的鸡数量
    maxv = (base - maxv) // 2
    # 更新堆
    heapq.heappush(a, (-maxv + base))

res = -sum(a) + n * k * m
print(res)
```
